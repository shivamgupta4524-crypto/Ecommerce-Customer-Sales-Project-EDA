```python
fig1, ax1 = plt.subplots(figsize=(15,10))

df.groupby(['sales_channel'], as_index=False)['product_cost'].sum().nlargest(10,'product_cost').sort_values('product_cost', ascending=False).plot(kind='bar', x='sales_channel', y='product_cost', ax=ax1)

plt.title('Each  product_cost')
plt.show()
```

```python
sns.set(rc={'figure.figsize': (15,10)})
x = df.groupby('sales_channel',as_index=False)['product_cost'].mean().sort_values(by='product_cost', ascending=False).head(10)
ax = sns.barplot(
    x='sales_channel',
    y='product_cost',
    data=x
)

for bars in ax.containers:
    ax.bar_label(bars)

plt.title('Each product_cost')
plt.show()
```

```python
df1 = df.groupby('sales_channel', as_index=False)['discount_amount'].mean().sort_values(by='discount_amount', ascending=False).head(10)

sns.set(rc={'figure.figsize': (15,10)})
sns.barplot(
    x='sales_channel',
    y='discount_amount',
    data=df1,
    hue='discount_amount',
    palette='coolwarm'
)
plt.title('Highest Discount Amount')
plt.show()
```

```python
month = df.groupby(df['order_date' ].dt.month) ['product_cost'].sum()
month.plot(marker='^')
plt.title(' month by product_cost')
plt.show()
```
